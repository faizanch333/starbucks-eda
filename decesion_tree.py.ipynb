```python
# ==========================================
# Load Dataset
# ==========================================

csv_path = None

for dirname, _, filenames in os.walk('/kaggle/input'):
    for filename in filenames:
        if filename.lower() == "starbucks.csv":
            csv_path = os.path.join(dirname, filename)

if csv_path is None:
    raise FileNotFoundError("starbucks.csv not found in /kaggle/input")

print("Dataset Found:")
print(csv_path)

df = pd.read_csv(csv_path)

print("\nDataset Loaded Successfully!")
```

```python
# ==========================================
# Display First and Last Records
# ==========================================

print("First Five Rows")
display(df.head())

print("\nLast Five Rows")
display(df.tail())
```

```python
# ==========================================
# Dataset Shape
# ==========================================

print("Rows and Columns")

print(df.shape)

print(f"\nNumber of Rows: {df.shape[0]}")
print(f"Number of Columns: {df.shape[1]}")
```

```python
# ==========================================
# Column Names
# ==========================================

print("Columns in Dataset:\n")

for column in df.columns:
    print(column)
```

```python
# ==========================================
# Dataset Information
# ==========================================

print(df.info())
```

```python
# ==========================================
# Missing Values
# ==========================================

missing = df.isnull().sum()

print(missing)

plt.figure(figsize=(8,5))
sns.heatmap(df.isnull(), cbar=False)

plt.title("Missing Values")
plt.show()
```

```python
# ==========================================
# Duplicate Rows
# ==========================================

duplicates = df.duplicated().sum()

print("Duplicate Rows:", duplicates)
```

```python
# ==========================================
# Descriptive Statistics
# ==========================================

display(df.describe())

print("\nIncluding Categorical Variables\n")

display(df.describe(include="all"))
```

```python
# ==========================================
# Numerical and Categorical Columns
# ==========================================

numerical_columns = df.select_dtypes(include=np.number).columns.tolist()

categorical_columns = df.select_dtypes(exclude=np.number).columns.tolist()

print("Numerical Columns\n")
print(numerical_columns)

print("\nCategorical Columns\n")
print(categorical_columns)
```

```python
# ==========================================
# Drink Category Distribution
# ==========================================

possible_category = None

for col in df.columns:
    if "category" in col.lower():
        possible_category = col
        break

if possible_category:

    plt.figure(figsize=(10,6))

    df[possible_category].value_counts().plot(
        kind="bar",
        color="steelblue"
    )

    plt.title("Distribution of Drink Categories")
    plt.xlabel(possible_category)
    plt.ylabel("Count")
    plt.xticks(rotation=45)

    plt.show()

else:
    print("No category column found.")
```

```python
# ==========================================
# Drink with Highest Calories
# ==========================================

calorie_col = None
drink_col = None

for col in df.columns:
    if "calorie" in col.lower():
        calorie_col = col

for col in df.columns:
    if any(word in col.lower() for word in ["drink","beverage","name","item","product"]):
        drink_col = col
        break

if calorie_col:

    highest = df.loc[df[calorie_col].idxmax()]

    print("Highest Calorie Drink\n")

    if drink_col:
        print(highest[[drink_col, calorie_col]])
    else:
        print(highest)

else:
    print("Calorie column not found.")
```

```python
# ==========================================
# Histogram
# ==========================================

if calorie_col:

    plt.figure(figsize=(8,5))

    sns.histplot(df[calorie_col], bins=20, kde=True)

    plt.title("Calorie Distribution")

    plt.show()

else:
    print("Calorie column not available.")
```

```python
# ==========================================
# Box Plot
# ==========================================

if calorie_col:

    plt.figure(figsize=(8,3))

    sns.boxplot(x=df[calorie_col])

    plt.title("Calories Box Plot")

    plt.show()

else:
    print("Calorie column not available.")
```

```python
# ==========================================
# Pie Chart of Categories
# ==========================================

if possible_category:

    df[possible_category].value_counts().plot(
        kind="pie",
        autopct="%1.1f%%",
        figsize=(7,7)
    )

    plt.ylabel("")
    plt.title("Drink Category Distribution")

    plt.show()

else:
    print("Category column not found.")
```

```python
# ==========================================
# Correlation Matrix
# ==========================================

nutrition_columns = []

keywords = [
    "calorie",
    "sugar",
    "fat",
    "caffeine",
    "protein",
    "carb",
    "fiber",
    "sodium"
]

for col in df.columns:
    for key in keywords:
        if key in col.lower():
            nutrition_columns.append(col)
            break

nutrition_columns = list(dict.fromkeys(nutrition_columns))

if len(nutrition_columns) >= 2:

    corr = df[nutrition_columns].corr(numeric_only=True)

    plt.figure(figsize=(8,6))

    sns.heatmap(
        corr,
        annot=True,
        cmap="coolwarm"
    )

    plt.title("Correlation Matrix")

    plt.show()

else:
    print("Not enough nutrition columns found.")
```

```python
# ==========================================
# Pair Plot
# ==========================================

if len(nutrition_columns) >= 2:

    sns.pairplot(df[nutrition_columns])

    plt.show()

else:
    print("Not enough nutrition columns.")
```

```python
# ==========================================
# Automatic Dataset Summary
# ==========================================

print("Dataset Summary")

print("- Total Rows:", df.shape[0])
print("- Total Columns:", df.shape[1])

print("- Numerical Columns:", len(numerical_columns))
print("- Categorical Columns:", len(categorical_columns))

print("- Missing Values:", df.isnull().sum().sum())

print("- Duplicate Rows:", df.duplicated().sum())

if calorie_col:
    print("- Maximum Calories:", df[calorie_col].max())
    print("- Average Calories:", round(df[calorie_col].mean(),2))
```
